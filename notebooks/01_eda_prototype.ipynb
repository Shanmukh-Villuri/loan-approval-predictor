# 01 — EDA + Prototyping (Home Credit application_train.csv)

This notebook does exploration only. All reusable logic lives in `src/` and is imported here.
Dataset: https://www.kaggle.com/competitions/home-credit-default-risk/data (`application_train.csv` only).
Place the file at `../data/application_train.csv` before running.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loading import load_application_train, minority_rate
from src.preprocessing import add_engineered_features, infer_column_types, build_preprocessor, encoded_feature_count

DATA = Path("../data/application_train.csv")
print("exists:", DATA.exists())
```

```python
# Load (use nrows=20000 for quick EDA on laptops)
df = load_application_train(DATA)  # or nrows=20000
print(df.shape, "minority_rate=", round(minority_rate(df['TARGET']), 4))
df[["TARGET", "AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "DAYS_BIRTH", "EXT_SOURCE_2"]].describe().T
```

```python
# Target imbalance + key distributions
fig, ax = plt.subplots(1, 2, figsize=(10, 3))
df["TARGET"].value_counts(normalize=True).plot.bar(ax=ax[0], title="TARGET rate (0=repay, 1=default)")
df["AMT_CREDIT"].hist(bins=50, ax=ax[1])
ax[1].set_title("AMT_CREDIT distribution")
plt.tight_layout(); plt.show()
```

```python
# Engineered features + post-encoding width (the ~250-300 check)
feat = add_engineered_features(df)
num, cat = infer_column_types(feat)
print(f"numeric={len(num)} categorical={len(cat)}")
pre = build_preprocessor(num, cat)
pre.fit(feat.drop(columns=['TARGET', 'SK_ID_CURR'], errors='ignore'))
print("encoded_feature_count=", encoded_feature_count(pre))
```
